## Gradient Descent

```python
* How your Linear Regression model finds out the Best Fit Line ?

Ans :- Gradient Descent
    >>> It is an optimization algorithm.
    >>> It is used to find optimal values of cost function(Evaluation Matric)
    >>> MSE is the default cost function in linear regression.
    >>> Best Fit Line:
        >> pass/goes through maximum number of datapoints
        >> It cant pass through all the datapoints
        >> It minimizes the distance between the datapoints and the line
        >> It will always pass through the point where x_mean and y_mean meets.
        >> Least sum of squares(OLS >> Ordinary Least Squares)
```

```python
>> Error/Residual >> difference between actual and predicted values

>> Loss Function >> Error for a single point
>> Cost Function >> all the errors combined

>> Sum of Squared Error/Residual Sum of Squares:
    sum((ya-yp)**2)

>> Mean Squared Error : sum((ya-yp)**2)/n

>> The lesser the errors, more reliable the model is

>> Variance:
    Stochastic ->  Uses one training example at a time. Fast, Noisy Updates, good for large datasets
    Deterministic -> Uses the entire dataset before updating parameters. Stable, slow updates, good for small datasets

>> When we use squared term in Cost Function >> Quadratic Equation:
    When we plot m & c against the error, it acquires a shape of bowl, we call it as convex fuction.
    - It gives us guarantee to have absolute minima(global minima)
    - Each point on the bowl have some m and c values and SSE which carries a line

>> Gradient Descent : Descenting down the gradient of bowl

    Learning step will become smaller and smaller as it descents down and should not oscillate/converge
    This step uses Bold Driver Algo in the Backend

    It internally used Partial Derivatives

>> Partial Derivatives:
    For a small change in m and c in either direction, how much the error drops or increases

Global Minima >> Best m and c , Lowest SSE/MSE

Local Minima >> Other lower grooves other than Global Minima(Deep Learning and Non COnvex Functions)
```

```python
Training Data
      │
      ▼
Guess m and c
      │
      ▼
Predict Y
      │
      ▼
Calculate Error (raw errors doesnt be goood to model so go for MSE over SSE (avg error over total error))
      │
      ▼
Compute MSE (Cost Function)
      │
      ▼
Use Partial Derivatives (to get to know which direction we move by changing m and c, if we are decreasing the error go with the flow, or increase means change direction)
      │
      ▼
Gradient Descent updates m and c
      │
      ▼
Repeat
      │
      ▼
Reach Global Minimum
      │
      ▼
Best Fit Line Ready


Q1. Why do we square the errors instead of using absolute values?

Because squaring:

Prevents positive and negative errors from cancelling out.
Penalizes large errors more heavily.
Produces a smooth, differentiable cost function, which makes Gradient Descent efficient.
Q2. Why does Linear Regression use Gradient Descent?

Gradient Descent iteratively updates the model parameters (coefficients and intercept) to minimize the cost function, usually the Mean Squared Error.

Q3. Why is the cost function convex in Linear Regression?

Because MSE is a quadratic function of the model parameters, resulting in a convex (bowl-shaped) surface with a single global minimum.


After calculating the error for each data point, we compute the Mean Squared Error (MSE).
We square the errors so that positive and negative errors do not cancel each other out, and so that larger errors are penalized more heavily than smaller ones.
Taking the mean gives the average error across all data points.
Gradient Descent then minimizes this MSE to find the best values of the model parameters (m and c), resulting in the Best Fit Line.

Interviewer: Why do we use MSE instead of simply summing the errors?

Answer:

We use MSE because summing raw errors can cause positive and negative errors to cancel each other out, giving a misleading impression of model performance.
Squaring the errors makes them all positive and penalizes larger errors more heavily.
Taking the mean provides an average error across all observations, which Gradient Descent can minimize efficiently.
```
