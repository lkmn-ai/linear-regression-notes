## Linear Regression Assumptions

```python
Assumptions of Linear Regression

Linear Regression works well only when the following assumptions are satisfied.
These assumptions are mainly based on the relationship between variables and the behaviour of the error (residuals).

---------------------------------------------------------

1) Linearity

Definition:
The model assumes that there is a linear relationship between the independent variable(s) and the dependent variable.

Meaning:
As the independent variable changes, the dependent variable should also change approximately in a straight-line pattern.

Example:
House Size ↑ → House Price ↑

Graph:

      *
    *
  *
*
---------------------------->

If the relationship is curved, Linear Regression may not perform well.

Why?
Because Linear Regression tries to fit only a straight line.

---------------------------------------------------------

2) Independence

Definition:
The observations should be independent of each other.

Meaning:
One observation should not influence another observation.

Example:
Marks scored by different students in an exam are usually independent.

Bad Example:
Today's stock price is influenced by yesterday's stock price.
(Time-series data often violates this assumption.)

Why?
Dependent observations can produce biased predictions and incorrect statistical conclusions.

Note:
Many people confuse this with "independent variables should be independent."
Actually, this assumption mainly refers to the independence of observations.

---------------------------------------------------------

3) No Multicollinearity

Definition:
Independent variables should not be highly correlated with each other.

Example:

House Area
Number of Rooms

These two variables are often highly correlated.

Problem:

House Area  --------\
                      ---> House Price
Rooms       --------/

Both predictors are providing almost the same information.

Why is this a problem?

The model becomes confused about which predictor is actually responsible for the prediction.

As a result:
- Coefficients become unstable.
- Interpretation becomes difficult.
- Model reliability decreases.

Solution:
- Remove one of the correlated variables.
- Use PCA or other feature selection techniques.

---------------------------------------------------------

4) Homoscedasticity

Definition:
The variance (spread) of residuals should remain constant for all predicted values.

Meaning:
Errors should have approximately the same spread throughout the model.

Good Graph:

      •
   •
 •
    •
 •
---------------------------->

Errors are evenly scattered.

Bad Graph (Heteroscedasticity):

•
  •
     •
          •
               •
---------------------------->

The spread increases as X increases.

Problem:
Unequal variance makes statistical tests and confidence intervals unreliable.

---------------------------------------------------------

5) Normality of Error Distribution

Definition:
The residuals (errors) should follow a Normal Distribution.

Graph:

           /\
          /  \
_________/    \_________

Mean = 0

Meaning:
Most errors should be close to zero.
Very large errors should occur rarely.

Why?

This assumption is mainly important for hypothesis testing and confidence intervals.

How to check?

- Histogram of residuals
- KDE Plot
- Q-Q Plot

---------------------------------------------------------

6) No Autocorrelation of Errors

Definition:
Residuals should not be correlated with each other.

Meaning:
The error of one observation should not depend on the previous error.

Good Example:

Errors:

2
-1
3
-2
1

Random pattern.

Bad Example:

2
3
4
5
6

Every error follows a pattern.

Example:
Weather prediction
Stock prices
Temperature forecasting

These datasets usually contain autocorrelation.

Problem:
The model misses some hidden pattern.

How to check?

Durbin-Watson Test

---------------------------------------------------------

7) Mean of Residuals = 0

Definition:
The average of all residuals should be zero.

Residual:

Residual = Actual - Predicted

Example:

Residuals:

2
-1
3
-2
-2

Mean

(2-1+3-2-2)/5 = 0

Why?

The Best Fit Line tries to pass through the centre of the data.

Some points lie above the line.
Some points lie below the line.

Positive and negative residuals balance each other.

If the average residual is not zero,
the fitted line is likely biased and is not the true Best Fit Line.

---------------------------------------------------------

Quick Revision

1. Linearity
Relationship between X and Y should be linear.

2. Independence
Observations should be independent of each other, st1 = 80, st2 = 90 but stock price influence tuesday market price on wednesday so that is problem

3. No Multicollinearity
Independent variables should not be highly correlated.

4. Homoscedasticity
Residual variance should remain constant, they should evenly spread throughout the model.

5. Normality
Residuals should follow a normal distribution, bell shaped curve, most errors should be near to zero (-1, 2, 1, -3, 2 and some is 20, 30, 50)

6. No Autocorrelation
Residuals should not have any pattern or correlation. any pattern means 1, 2, 3, 4, 5 it should not be like this

7. Mean of Residuals = 0
Average residual should be zero. if not it will be like biased.
```

```python
Variance Inflation Factor (VIF)

What is VIF?

Variance Inflation Factor (VIF) is a metric used to detect Multicollinearity among the independent variables.

Why do we use VIF?

If two or more independent variables are highly correlated, the model becomes confused about which variable is actually influencing the target variable.

VIF helps us identify such highly correlated features.

Formula:

VIF = 1 / (1 - R²)

Here,
R² is obtained by regressing one independent variable against all the other independent variables.

Interpretation:

VIF = 1
- No correlation with other independent variables.
- Best case.

VIF = 1 to 5
- Low correlation.
- Usually acceptable.

VIF = 5 to 10
- Moderate correlation.
- Keep only if the feature is important.

VIF > 10
- High Multicollinearity.
- Generally recommended to remove or combine the feature.

Example:

Suppose we have

Area
Rooms
House Age

If

Area and Rooms are highly correlated,

Area ↑ → Rooms ↑

Then VIF for both Area and Rooms will be high.

This indicates Multicollinearity.

---------------------------------------------------------

Ordinary Least Squares (OLS)

What is OLS?

Ordinary Least Squares (OLS) is a statistical method used to find the Best Fit Line.

Objective:

It minimizes the Sum of Squared Errors (SSE) between the actual values and the predicted values.

Formula:

SSE = Σ(Actual - Predicted)²

Why square the errors?

- Prevents positive and negative errors from cancelling each other.
- Penalizes larger errors more.
- Produces a convex (bowl-shaped) cost function.

Result:

OLS finds the best values of

- Slope (m)
- Intercept (c)

so that the prediction error is minimum.

---------------------------------------------------------

Stochastic Gradient Descent (SGD)

Problem with Batch Gradient Descent

Suppose we have

100 observations
10 features

Batch Gradient Descent calculates gradients using all 100 observations before updating the model parameters.

For every iteration, all observations are processed.

If we train for 1000 iterations,

100 observations × 1000 iterations = 100,000 observations processed.

Now imagine a dataset with

10 million observations.

Batch Gradient Descent becomes very slow.

---------------------------------------------------------

What is SGD?

Stochastic means Random.

Instead of using the entire dataset,

SGD randomly picks one training example at a time to update the model parameters.

Process:

Pick one observation
↓
Calculate prediction
↓
Calculate error
↓
Update m and c immediately
↓
Pick another random observation
↓
Repeat

---------------------------------------------------------

Advantages of SGD

- Much faster for large datasets.
- Requires less memory.
- Can train on streaming or continuously arriving data.
- Suitable for very large Machine Learning problems.

---------------------------------------------------------

Disadvantages of SGD

- Updates are noisy because only one sample is used.
- The cost function may fluctuate before converging.
- May take more iterations to reach the minimum.

---------------------------------------------------------

Mini-Batch Gradient Descent

Instead of using

1 observation (SGD)

or

Entire dataset (Batch Gradient Descent),

Mini-Batch Gradient Descent uses a small batch of observations.

Example:

Batch Size = 32

Every update uses only 32 observations.

Advantages:

- Faster than Batch Gradient Descent.
- More stable than SGD.
- Most commonly used optimization technique in Deep Learning.

---------------------------------------------------------

Quick Revision

Batch Gradient Descent
- Uses the complete dataset.
- Slow.
- Stable updates.

Stochastic Gradient Descent (SGD)
- Uses one random observation.
- Very fast.
- Noisy updates.

Mini-Batch Gradient Descent
- Uses a small batch (32, 64, 128, etc.).
- Fast and stable.
- Most widely used in practice.
```

\## Evaluation Metric for Regression

```python
Model Evaluation Metrics for Linear Regression

A good regression model should have:

High Values:
1. R² Score
2. Adjusted R² Score

Low Values:
1. MAE (Mean Absolute Error)
2. MSE (Mean Squared Error)
3. RMSE (Root Mean Squared Error)

---------------------------------------------------------

Loss Function / Cost Function

A Cost Function is a mathematical function that measures how far the predicted values are from the actual values.

Gradient Descent tries to minimize (optimize) the cost function by finding the best values of the model parameters.

Lower Cost Function → Better Model

---------------------------------------------------------

1) Mean Absolute Error (MAE)

Definition:

MAE is the average of the absolute differences between the actual values and the predicted values.

Formula:

MAE = Σ |Actual - Predicted| / n

Example:

Actual = 10, 20, 30
Predicted = 12, 18, 35

Errors:
2, 2, 5

MAE = (2 + 2 + 5) / 3
     = 3

Advantages:
- Easy to understand.
- Treats all errors equally.
- Less sensitive (more robust) to outliers.

Disadvantages:
- Does not penalize large errors heavily.

---------------------------------------------------------

2) Mean Squared Error (MSE)

Definition:

MSE is the average of the squared differences between the actual values and the predicted values.

Formula:

MSE = Σ (Actual - Predicted)² / n

Example:

Errors:
2, 2, 5

Squared Errors:
4, 4, 25

MSE = (4 + 4 + 25) / 3
    = 11

Advantages:
- Most commonly used loss function in Linear Regression.
- Penalizes large errors more heavily.
- Smooth and differentiable, making it suitable for Gradient Descent.

Disadvantages:
- Highly sensitive to outliers.

---------------------------------------------------------

3) Root Mean Squared Error (RMSE)

Definition:

RMSE is the square root of the Mean Squared Error.

Formula:

RMSE = √MSE

or

RMSE = √(Σ(Actual - Predicted)² / n)

Advantages:
- Same unit as the target variable.
- Easier to interpret than MSE.
- Penalizes large errors.

Disadvantages:
- Sensitive to outliers.
- Not as robust as MAE.

Applications:
- Linear Regression
- Time Series Forecasting
- Deep Learning Regression Problems

---------------------------------------------------------

Comparison

MAE
- Uses absolute error.
- Less sensitive to outliers.
- Easy to interpret.

MSE
- Uses squared error.
- Highly sensitive to outliers.
- Most commonly used optimization (loss) function.

RMSE
- Square root of MSE.
- Same unit as the target variable.
- Easier to explain to business users.

---------------------------------------------------------

Which Metric Should We Use?

If the dataset contains many outliers:
→ Prefer MAE

If we want to penalize large errors:
→ Prefer MSE

If we want an error metric in the same unit as the target variable:
→ Prefer RMSE

---------------------------------------------------------

Quick Revision

Higher is Better:
- R² Score
- Adjusted R² Score

Lower is Better:
- MAE
- MSE
- RMSE
```
