## Overfitting and Underfitting

```python
Overfitting:
    Train Data Accuracy >> high >> 90%
    Test Data Accuracy  >> low  >> 63%

Underfitting:
    Train Data Accuracy >> low >> 62%
    Test Data Accuracy  >> low  >> 59%

Best Model:
    Train Data Accuracy >> high >> 90%
    Test Data Accuracy  >> high >> 88%
```

## Bias and Variance

```python
Bias: loss function
    - difference between actual and predicted values(residual) of training dataset
    - high bias:
        - Not capturing training data >> Training Accuracy >> low
        - 60%, 55%

    - low bias:
        - Able to capture training data >> Training Accuracy >>> high
        - 94%, 89 %

Variance: cost function
    - difference between two different dataset accuracies(training and testing) (Performance of model)
    - High Variance:
        Case_1:
            Train Data Accuracy >> high >> 90%
            Test Data Accuracy >> low >> 63%

        Case_2:(Impossible)
            Train Data Accuracy >> low >> 63%
            Test Data Accuracy >> high >> 90%

    - Low Variance:
        Case_1:
            Train Data Accuracy >> high >> 90%
            Test Data Accuracy >> high >> 91%

        Case_2:
            Train Data Accuracy >> low >> 63%
            Test Data Accuracy >> low >> 59%


* Worst Model:
    High Bias and High Variance

* Underfitted MOdel:
    High Bias and Low Variance

* Overfitted Model:
    Low Bias and High Variance

* Best Model :
    Low Bias and Low Variance

ML model main goal is to generalise our model

Generelization:
    Models ability to adapt to the unseen data by taking input from it
    and predict a good output.
```

```python
Variance :
    train_accuracy - test_accuracy = less than equals 5 (Optimal)  >> Low Variannce

    train_accuracy - test_accuracy = greater than 10 (need to resolve) >> High Variance
```

```python
Noise >> It is unnecessary or irrelevant data that reduces the performance of the model
```

## How to avoid Overfitting

```python
1. Hyperparameter Tuning:
    GridSearchCV
    RandomizedSearchCV

2. Cross Validations:
    K-FOld Cross Validation
    StratifiedKFold
    cross_val_score

3. Train with more dataset:(new rows and sampling)
    5000 > 6000
    8000 > 10000

4. Pruning: (Cutting the branches of trees)
    CART(Classification and Regression Trees) >> DT, RF, AB

5. Reduce the number of features:
    Use proper feature selection(Dimensionity Reduction TEchnique)

6. Early Stopping(Deep LEarning)

7. Regularization:
    1. L1 Regularization >> Lasso Regression >> Feature Selection
    2. L2 Regularization >> Ridge Regression

L1 Regularization (Lasso): L1 can shrink some coefficients all the way to zero.

L2 Regularization (Ridge): Ridge also penalizes large coefficients, but instead of making them exactly zero, it usually shrinks them.


8. Ensemble Techniques:
    1. Bagging:
        Random Forest(Forest of Trees)

    2. Boosting:
        1. Adaboost
        2. Gradient Descent
        3. XGBoost


| Technique             | Main Idea                                      | Used For                             |
| --------------------- | ---------------------------------------------- | ------------------------------------ |
| Hyperparameter Tuning | Find the best model settings                   | Better performance, less overfitting |
| GridSearchCV          | Try all parameter combinations                 | Small search spaces                  |
| RandomizedSearchCV    | Try random combinations                        | Large search spaces                  |
| K-Fold CV             | Evaluate on multiple train/test splits         | Reliable evaluation                  |
| Stratified K-Fold     | Preserve class distribution                    | Classification                       |
| More Data             | Learn real patterns instead of memorizing      | Reduce overfitting                   |
| Pruning               | Cut unnecessary tree branches                  | Decision Trees                       |
| Feature Selection     | Remove irrelevant features                     | Simpler models                       |
| Early Stopping        | Stop training before memorization              | Deep Learning                        |
| L1 (Lasso)            | Shrinks some coefficients to zero              | Feature selection                    |
| L2 (Ridge)            | Shrinks coefficients without removing features | Reduce overfitting                   |
| Bagging               | Train models independently and combine them    | Reduce variance                      |
| Boosting              | Train models sequentially to correct errors    | Reduce bias and improve accuracy     |
```

## How to avoid Underfitting

```python
1. Incease number of features:
    1. Adding New Features
    2. Add Derived Features(Feature Engineering)

2. Hyperparameters Tuning
3. Use Correlated Features
4. Remove Outliers
5. Increase Data Size
6. Handling Missing Values(Mean, Median, Mode)
```
