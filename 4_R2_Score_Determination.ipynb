```python
>> Errors:
    Residual Errors(SSE/RSS) = actual - pred >> Unexplained Error
    Regression Errors (SSR)  = pred - mean   >> Explained Error
    Total Errors (SST)       == Residual + Regression
                                actual - mean

# Coefficient of Determination (R² Score)

What is R² Score?

The Coefficient of Determination (R² Score) is an evaluation metric used in Linear Regression to measure how well the model fits the data.

Uses:
- Evaluate the performance of a regression model.
- Measure the goodness of fit of the Best Fit Line (BFL).
- Determine how much variation in the dependent variable is explained by the independent variable(s).
- Check the reliability of the model.

Interpretation:
- Higher R² → Better model.
- Lower unexplained variance (residual error) → Better model.
- Higher R² means the model explains more variation in the target variable.

Formula:

R² = 1 - (SSE / SST)

Where:
SSE (Sum of Squared Errors) = Unexplained variation (Residual Error)
SST (Total Sum of Squares) = Total variation present in the actual data.

Range of R²:
Normally,
0 ≤ R² ≤ 1

However, R² can also become negative when the model performs worse than predicting the mean of the target variable.

Cases:

Case 1: Perfect Model
When:
SSE = 0

Then:
R² = 1

Meaning:
- Perfect prediction.
- No residual error.
- Best possible model.

Case 2: Good Model
When:
SST > SSE

Then:
0 < R² < 1

Meaning:
- The model explains a significant portion of the variance.
- Better fit.

Case 3: Poor Model
When:
SSE > SST

Then:
R² < 0

Meaning:
- The model performs worse than simply predicting the average (mean).
- Very poor model.

Single vs Multiple Independent Variables

If there is only one independent variable:
R² = r × r
where r is the Pearson Correlation Coefficient.

If there are multiple independent variables:
It is called the Coefficient of Multiple Determination.

Limitation of R²:

R² is not always considered the best evaluation metric because:
- It never decreases when new predictors are added.
- Even irrelevant (bad) predictors can increase or maintain the R² score.
- This may give a false impression that the model has improved.

Adjusted R² Score

Adjusted R² is a modified version of R².

Advantages:
- It considers both the number of predictors and the number of observations.
- It increases only when a newly added predictor actually improves the model.
- It decreases if the new predictor is irrelevant.
- It is more reliable than R² for Multiple Linear Regression.

Relationship:
Adjusted R² ≤ R²

If a good predictor is added:
Adjusted R² increases.

If a bad predictor is added:
Adjusted R² decreases.

# Quick Revision Table

| Metric          | Purpose                                                                                 |
| --------------- | --------------------------------------------------------------------------------------- |
| **R² Score**    | Measures how well the regression model explains the variability in the target variable. |
| **Formula**     | (R^2 = 1 - \frac{SSE}{SST})                                                             |
| **Best Value**  | 1                                                                                       |
| **Worst Value** | Can be negative                                                                         |
| **R² = 1**      | Perfect model                                                                           |
| **0 < R² < 1**  | Good model                                                                              |
| **R² < 0**      | Worse than predicting the mean                                                          |
| **Limitation**  | Increases even when irrelevant features are added                                       |
| **Adjusted R²** | Penalizes unnecessary predictors and is more reliable for multiple regression           |

---

## Interview Question

Q: Why do we use Adjusted R² instead of R²?**

Answer:

> R² always stays the same or increases when additional predictors are added, even if those predictors are not useful.
Adjusted R² penalizes unnecessary predictors and increases only when the new predictor meaningfully improves the model.
Therefore, Adjusted R² is preferred for evaluating multiple linear regression models.
```
