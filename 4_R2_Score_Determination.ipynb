## Linear Regression

```python
>> Errors:
    Residual Errors(SSE/RSS) = actual - pred >> Unexplained Error
    Regression Errors (SSR)  = pred - mean   >> Explained Error
    Total Errors (SST)       == Residual + Regression
                                actual - mean

>> Coefficient of Determination(r2_score):
    >> To check the Model/to check the reliability of the model.
    >> It is used to check the goodness of BFL
    >> Evaluate the performance of the model/it will define how the model is
    >> Lesser the unexplained variance/residual, Better the model
    >> If we have 1 ind var:
        >> Coefficient of Determination
        >> R * R
    >> If we have more than 1 ind var:
        >> Coefficient of Multiple Determination

    >> r2_score ranges from 0 to 1

    >> r2_score = 1 - SSE/SST

    >> When SSE = 0:
        r2_score == 1

    >> When SST > SSE:
        r2_score >> 0 to 1

    >> When SSE > SST:
        r2_score = -ve

    >> r2_score is not considered as very good metric to evaluatre the model:
        Beacuase it gets impacted by the bad predictors

>> So we make use of Adjusted R2_score:
    >> Modified version of r2_score
    >> It will only increase when we add good predictors
    >> Adjusted_r2 score will always be less than r2_score
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

```python
Adjusted_r2_Score = 1 - (((1-r2_score)*(n-1))/(n-k-1))
n = no. of datapoints
k = no. of ind var
r2_score >> coeff of determination
```
