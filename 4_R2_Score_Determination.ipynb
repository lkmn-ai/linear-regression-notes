````python
>> Errors:
    Residual Errors(SSE/RSS) = actual - pred >> Unexplained Error
    Regression Errors (SSR)  = pred - mean   >> Explained Error
    Total Errors (SST)       == Residual + Regression
                                actual - mean

Sure! Here's a **well-structured and interview-friendly** version of your notes.

---

# Coefficient of Determination (R² Score)

## What is R² Score?

The **Coefficient of Determination (R² Score)** is an evaluation metric used in **Regression** to measure **how well the Linear Regression model fits the data**.

### It is used to:

* Evaluate the performance of a regression model.
* Measure the **goodness of fit** of the Best Fit Line (BFL).
* Determine how much of the variation in the dependent variable is explained by the independent variable(s).
* Check the reliability of the model.

---

# Interpretation

* **Higher R²** → Better model.
* **Lower unexplained variance (residual error)** → Better model.
* A higher R² indicates that the model explains a larger portion of the variability in the target variable.

---

# Formula

[
R^2 = 1 - \frac{SSE}{SST}
]

Where:

* **SSE (Sum of Squared Errors)** = Unexplained variation (prediction error)
* **SST (Total Sum of Squares)** = Total variation present in the actual data

---

# Range of R² Score

Normally,

[
0 \le R^2 \le 1
]

However, **R² can also be negative** if the model performs worse than simply predicting the mean.

---

# Different Cases

### Case 1: Perfect Model

When

```text
SSE = 0
```

Then

```text
R² = 1
```

Meaning:

* Perfect prediction
* No prediction error
* Best possible model

---

### Case 2: Good Model

When

```text
SST > SSE
```

Then

```text
0 < R² < 1
```

Meaning:

* The model explains a significant portion of the data.
* Better fit.

---

### Case 3: Poor Model

When

```text
SSE > SST
```

Then

```text
R² < 0
```

Meaning:

* The model performs worse than simply predicting the average (mean) of the target variable.
* Very poor model.

---

# One vs Multiple Independent Variables

### Single Independent Variable

The coefficient of determination is simply

[
R^2 = r \times r
]

where **r** is the Pearson Correlation Coefficient.

---

### Multiple Independent Variables

When there are multiple independent variables,

it is called the **Coefficient of Multiple Determination**.

The calculation is still reported as an **R² Score**, but it is obtained from the multiple regression model rather than simply squaring a single correlation coefficient.

---

# Limitation of R² Score

R² is **not always the best evaluation metric** because:

* It **never decreases** when new features are added.
* Even **irrelevant (bad) predictors** can increase or maintain the R² score.
* Therefore, it may falsely indicate that the model has improved.

---

# Adjusted R² Score

To overcome this limitation, we use **Adjusted R²**.

### What is Adjusted R²?

Adjusted R² is a modified version of R² that considers:

* Number of predictors (features)
* Number of observations

It penalizes unnecessary features.

---

# Advantages of Adjusted R²

* It increases **only when a newly added predictor genuinely improves the model**.
* It decreases if the added predictor is irrelevant.
* It provides a more reliable measure for models with multiple independent variables.

---

# Relationship Between R² and Adjusted R²

Generally,

```text
Adjusted R² ≤ R²
```

* If the new feature is useful → Adjusted R² increases.
* If the new feature is not useful → Adjusted R² decreases.

---

# Quick Revision Table

| Metric          | Purpose                                                                                 |
| --------------- | --------------------------------------------------------------------------------------- |
| **R² Score**    | Measures how well the regression model explains the variability in the target variable. |
| **Formula**     | (R^2 = 1 - \frac{SSE}{SST})                                                             |
| **Best Value**  | 1                                                                                       |
| **Worst Value** | Can be negative                                                                         |
| **R² = 1**      | Perfect model                                                                           |
| **0 < R² < 1**  | Good model                                                                              |
| **R² < 0**      | Worse than predicting the mean                                                          |
| **Limitation**  | Increases even when irrelevant features are added                                       |
| **Adjusted R²** | Penalizes unnecessary predictors and is more reliable for multiple regression           |

---

## Interview Question

**Q: Why do we use Adjusted R² instead of R²?**

**Answer:**

> R² always stays the same or increases when additional predictors are added, even if those predictors are not useful. Adjusted R² penalizes unnecessary predictors and increases only when the new predictor meaningfully improves the model. Therefore, Adjusted R² is preferred for evaluating multiple linear regression models.
````
