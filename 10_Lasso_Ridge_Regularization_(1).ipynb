## Regularization

```python
1. Lasso Regression/ Lasso Regularization/ L1 Norm(L1 Regularization)
2. Ridge Regression/ Ridge Regularization/ L2 Norm(L2 Regularization)

Overfitting:
    Low Bias and High Variance

    Training Accuracy >> high
    Testing Accuracy  >> low

To reduce Overfitting:
    We have to reduce Variance

Regularization:
    It is a technique to prevent the model from overfitting by adding extra information to it.
    or by penelising the coeff of variables.

    This technique is used to caliberate ML Model in order to optimise cost function

    Regularization works by penelising the coeff/magnitude of features along with optimising the error

lambda/alpha >> Ranges from 0 onwards
```

## Ridge Regularization(L2 Regularization)

```python
L2 Regularization:
    >> It performs by shrinking the coefficient
    >> It modifies the overfitted or underfitted model by
    adding the penalty to the sum of square of magnitude of features
    Suppose if there is large coefficient then it will be penalized, coz the answer becomes more huge and more complex.


    Instead of minimizing only SSE,
    SSE = sum(act - pred)**2 = Cost Function
    Cost Function = SSE + alpha*(slope)^2 (Sum of square of magnitude)
    alpha*(slope)^2 >> Penalty Term

    It reduce the problem of Multicolinearity
    It shrinks the values of magnitude of feature almost near to zero
```

## Lasso Regularization(L1 Regularization)

```python
L1 Regularization:
    SSE = sum(act - pred)**2 = Cost Function
    Cost Function = SSE + alpha*|slope| (Sum of magnitude)
    alpha*|slope| >> Penalty Term

    It shrinks the values of magnitude of feature almost equal to zero
    Lasso is also used for Feature Selection >> Embedded Method(Feature Slection)
```
