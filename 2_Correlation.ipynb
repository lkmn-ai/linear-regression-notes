```python
Q1. Why use a scatter plot?

To visually identify the relationship between two continuous variables and see whether it is positive, negative, or has no clear trend.

Q2. Why use a correlation matrix?

To measure the strength and direction of linear relationships between all pairs of numeric features, helping identify useful predictors and multicollinearity.

Q3. Why drop Species before calling corr()?

Because Pearson correlation operates on numeric data; Species is a categorical text column.

Q4. Why does the diagonal of the correlation matrix always contain 1?

Because every variable is perfectly correlated with itself.

Q5. What is the difference between a scatter plot and a heatmap?
Scatter plot: Shows the relationship between two variables.
Heatmap: Summarizes the correlations between all numeric variables in one visualization.
```
