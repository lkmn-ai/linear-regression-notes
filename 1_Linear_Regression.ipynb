## Linear Regression

```python
* Linear >> In a straight line or a straight path
* Regression >> Predicting continous values or a real number

* Linear Regression :
    It is a predictive model used to find out the linear relationship between a dependent var and
    one or more independent var and predicts values.

* Target Columns: (Continous values)
    Price, Sales, Stock Price, BP, Temp

* It is a Parametric Algorithm:
    Parametric Algorithm >> Assumptions of distribution of data

* Types of Linear Regression:
    1. Simple Linear Regression:
        Only one Independent Variable
        y = mx + c
        y = Dependent Var
        m = slope
        x = Independent Var
        c = Constant/Intercept

    2. Multiple Linear Regression:
        Two or more Independent Variables
        y = m1x1 + m2x2 + ........ + mNxN + c
        y = Dependent Var
        m1,m1,...... = slope
        x1,x2,...... = Independent Vars
        c = Constant/Intercept

Dependent Var >> Continous/Numeric
Independent Var >> Continous/Numeric/ discrete

Aim : To predict a Dependent Var (Y) base on Independent Var (X)

* The main target is to fing out Best FIt Line(BFL)
* The Best Fit Line is also known as Regression Line/Linear Model
```

```python
m and c reflects the relationship between x and y in your dataset
m and c will form us the BFL/model

c >> intercept >> What is the value of y, when x = 0.
m >> theta >> When we change 1 unit in x, how much your y changes
```

```python
* When we have multiple features

* Features >> Dimensions in Feature space
* Feature space >> Total no of dimensions in data
* Model dimensions >> no of independent var
```

```python
* Assumption of Independence:
    Algorithm assumes that ind var are independent to each other
    They do not influence each other

* In a real world ind var influence each other

* But alorithm assumes that they are not related

* This is a problem >> Problem of Colinearity
>> So when we have strong Colinearity, We need to resolve
```

## Coefficient of Correlation- (R Value)

```python
>> It indicates how closely the relationship (Linear) of two variables is.
>> The relationship between two variables:
    - ind with ind var
    - dependent with indepedent
>> By defualt we use Pearson Correlation Coefficient, it ranges from -1 to +1
>> When x increases, y also increases >> positive corr >> Good Predictor with r value close to +1
>> When x increases, y decreases >> negative corr >> Good Predictor with r value close to -1
>> When corr gives values very close to -1 or +1 >> indicates linear relation between two variables
>> When there is no linear relation between 2 var >> Bad Predictors

>> When R is -1 or +1 >> Ideal relation >> this doesnt exist >> BFL passes through all the datapoints:
    x = y

>> When R ranges from:
    Good Predictors:
        + 0.7 to 1
        - -0.7 to -1

    Bad Predictors:
        - -0.3 to 0.3

    0.3 to 0.7 and -0.3 to -0.7:
        We need to look/indepth/research about these features

>> R value is near to 0 >> When datapoints are scattered
>> R value is near to +1 >> When x and y shows positive linear relation
>> R value is near to -1 >> When x and y shows neagative linear relation
```

## Computation of Correlation (Formula)

```python
>> Variance >> It measures the variability from mean
    Var >> sum((Xi-X^)**2)/n

>> Central Tendency:
    The statistical measure that identifies a single value as a representative of entire distribution.
    It aims to provide an accurate description of entire data
    Mean , Median and Mode

>> Variance gives reliability of Central Values

>> Covariance:
    How two features vary together/how they influence together

    COv = ((Xi-X^)(Yi-Y^))/n

>> if Covariance == 0, Then Correlation == 0

>> Standard Deviation:
    std = np.sqrt(sum((Xi-X^)**2)/n)

>> Pearson Correlation Coefficient:
    R(x,y) = sum((Xi-X^)(Yi-Y^)) / np.sqrt(sum(Xi-X^)**2)* np.sqrt(sum(Yi-Y^)**2)
            COv/std(x)*std(y)

Spearmans Rank Correlation Coefficient
Kendall tau Correlation Coefficient
```
